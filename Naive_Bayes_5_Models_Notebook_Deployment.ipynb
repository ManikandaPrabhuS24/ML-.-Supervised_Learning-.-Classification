```python
## 18. Loading the Saved Model
import pickle
import pandas as pd

filename = "Naive_Bayes_5_Models.sav"
load_model = pickle.load(open(filename, 'rb'))
print("Model loaded successfully.")
```

```python
## 19. Taking New Customer Input
age = float(input("Age: "))
estimated_salary = float(input("Estimated Salary: "))
gender_male = int(input("Gender Male (0 or 1): "))
new_customer = [[age, estimated_salary, gender_male]]
```

```python
## 20. Generating the Purchase Prediction
prediction = load_model.predict(new_customer)
if prediction[0] == 0:
    print("Not Purchased")
else:
    print("Purchased")
```

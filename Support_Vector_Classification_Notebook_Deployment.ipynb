```python
## 16. Loading the Saved Model
import pickle
import pandas as pd
from sklearn.preprocessing import StandardScaler
s = StandardScaler()
load_model = pickle.load(open("Finalized_model.sav", 'rb'))

## 17.Taking User Input (Converting Input to DataFrame)
age = int(input("Enter the Age : "))
Est_Salary = int(input("Enter the Estimated Salary : "))
gender_male = int(input("Enter the Gender (0/1) : "))
dataset = pd.DataFrame([[age, Est_Salary, gender_male]], columns = [ "Age", "EstimatedSalary", "Gender_Male"])

## 18. Applying Feature Scaling
input_Scaled = s.fit_transform(dataset)
r = load_model.predict(input_Scaled)
r = r.item()

## 19. Displaying the Final Prediction
print("The Prediction for Social Network Ads Purchased or Not is ",r)
```

```python
if r == 0:
    print("Not Purchased")
else:
    print("Purchased")
```

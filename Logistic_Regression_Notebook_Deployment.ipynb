```python
## 14. Loading the Saved Model
import pickle
import pandas as pd
load_model = pickle.load(open("Finalized_model.sav", 'rb'))

## 15. Making Predictions Using the Deployed Model
age = int(input("Enter the Age : "))
Est_Salary = int(input("Enter the Estimated Salary : "))
gender_male = int(input("Enter the Gender (0/1) : "))
dataset = pd.DataFrame([[age, Est_Salary, gender_male]], columns = [ "Age", "EstimatedSalary", "Gender_Male"])
r = load_model.predict(dataset)
r = r.item()

## 16. Displaying the Final Prediction
print("The Prediction for Social Network Ads Purchased or Not is ",r)
```

```python
if r == 0:
    print("Not Purchased")
else:
    print("Purchased")
```

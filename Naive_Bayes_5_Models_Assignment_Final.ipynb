```python
# Naive Bayes Classification Using Five Naive Bayes Models
# Author: MANIKANDAPRABHU.S
# Objective: Compare five Naive Bayes classifiers for predicting whether a customer will purchase a product.
```

```python
## 1. Import Libraries
import pandas as pd
```

```python
## 2. Loading the Dataset
dataset=pd.read_csv("Social_Network_Ads.csv")
```

```python
## 3. Exploring the Dataset
dataset
```

```python
## 4. Encoding Categorical Variables
dataset=pd.get_dummies(dataset,dtype=int,drop_first=True)
```

```python
## 5. Removing Unnecessary Columns
dataset=dataset.drop("User ID",axis=1)
```

```python
## 6. Checking the Prepared Dataset
dataset
```

```python
## 7. Checking Dataset Columns
dataset.columns
```

```python
## 8. Preparing Features and Target Variable
indep=dataset[['Age', 'EstimatedSalary',  'Gender_Male']]
dep=dataset["Purchased"]
```

```python
## 9. Splitting Dataset into Training and Testing Sets
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(indep,dep,test_size=1/3,random_state=0)
```

```python
## 10. Training Categorical Naive Bayes
from sklearn.naive_bayes import CategoricalNB
classifier1=CategoricalNB()
classifier1.fit(x_train,y_train)
y_pred=classifier1.predict(x_test)
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test,y_pred)
from sklearn.metrics import  classification_report
clf_report= classification_report(y_test,y_pred)
print(cm)
print(clf_report)
```

```python
## 11. Training Bernoulli Naive Bayes
from sklearn.naive_bayes import BernoulliNB
classifier2=BernoulliNB()
classifier2.fit(x_train,y_train)
y_pred=classifier2.predict(x_test)
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test,y_pred)
from sklearn.metrics import  classification_report
clf_report= classification_report(y_test,y_pred)
print(cm)
print(clf_report)
```

```python
## 12. Training Gaussian Naive Bayes
from sklearn.naive_bayes import GaussianNB
classifier3=GaussianNB()
classifier3.fit(x_train,y_train)
y_pred=classifier3.predict(x_test)
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test,y_pred)
from sklearn.metrics import  classification_report
clf_report= classification_report(y_test,y_pred)
print(cm)
print(clf_report)
```

```python
## 13. Training Multinomial Naive Bayes
from sklearn.naive_bayes import MultinomialNB
classifier4=MultinomialNB()
classifier4.fit(x_train,y_train)
y_pred=classifier4.predict(x_test)
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test,y_pred)
from sklearn.metrics import  classification_report
clf_report= classification_report(y_test,y_pred)
print(cm)
print(clf_report)
```

```python
## 14. Training Complement Naive Bayes
from sklearn.naive_bayes import ComplementNB
classifier5=ComplementNB()
classifier5.fit(x_train,y_train)
y_pred=classifier5.predict(x_test)
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test,y_pred)
from sklearn.metrics import  classification_report
clf_report= classification_report(y_test,y_pred)
print(cm)
print(clf_report)
```

```python
## 15. Data Visualization
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6,4))
sns.countplot(data=dataset, x="Purchased")
plt.title("Purchase Distribution")
plt.xlabel("Purchased")
plt.ylabel("Count")
plt.show()

plt.figure(figsize=(8,6))
sns.scatterplot(data=dataset, x="Age", y="EstimatedSalary", hue="Purchased", s=80)
plt.title("Age vs Estimated Salary by Purchase")
plt.xlabel("Age")
plt.ylabel("Estimated Salary")
plt.show()

plt.figure(figsize=(7,5))
sns.heatmap(dataset.corr(numeric_only=True), annot=True, fmt=".2f")
plt.title("Feature Correlation Heatmap")
plt.show()
```
